```python
# Importação das bibliotecas necessárias para manipulação de dados, modelagem e visualização
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

```

```python


# 1. PREPARAÇÃO / GERAÇÃO DOS DADOS
# Finalidade: Criar um conjunto de dados sintético com relação linear (y = 4 + 3x + ruído)
# para simular o treinamento do modelo.

np.random.seed(42)  # Define semente para reprodutibilidade dos resultados
X = 2 * np.random.rand(100, 1)  # Variável independente (Feature/Atributo)
y = (
    4 + 3 * X + np.random.randn(100, 1)
)  # Variável dependente (Target/Rótulo)

```

```python

# 2. SELEÇÃO E TREINAMENTO DO MODELO
# Finalidade: Instanciar o algoritmo de Regressão Linear e ajustar (fit) os parâmetros
# do modelo (intercepto e inclinação) aos dados de treino.

model = LinearRegression()  # Instancia o modelo linear[cite: 1]
model.fit(X, y)  # Treina o modelo encontrando os coeficientes[cite: 1]

# Exibe os parâmetros aprendidos pelo modelo
print(f"Intercepto (b0): {model.intercept_[0]:.4f}")  # Ponto onde a reta cruza o eixo Y[cite: 1]
print(f"Coeficiente Angular (b1): {model.coef_[0][0]:.4f}")  # Inclinação da reta[cite: 1]

```

```python

# 3. REALIZAÇÃO DE PREDIÇÕES
# Finalidade: Utilizar o modelo treinado para prever novos valores de 'y' com base em 'X'[cite: 1].

X_new = np.array([[1.5]])  # Novo dado de entrada[cite: 1]
y_predict = model.predict(X_new)  # Faz a predição[cite: 1]
print(f"Previsão para X = 1.5: {y_predict[0][0]:.4f}")

```

```python

# 4. AVALIAÇÃO DO MODELO
# Finalidade: Calcular a métrica de erro RMSE (Raiz do Erro Quadrático Médio)
# para quantificar a precisão do modelo no conjunto de dados[cite: 1].

y_pred_all = model.predict(X)  # Predições para todo o dataset[cite: 1]
mse = mean_squared_error(y, y_pred_all)  # Calcula o Erro Quadrático Médio[cite: 1]
rmse = np.sqrt(mse)  # Calcula a Raiz do MSE (RMSE)[cite: 1]
print(f"RMSE do Modelo: {rmse:.4f}")

```

```python

# 5. VISUALIZAÇÃO DOS RESULTADOS
# Finalidade: Plotar os pontos reais de dados e a reta de regressão ajustada[cite: 1].

plt.scatter(
    X, y, color="blue", alpha=0.6, label="Dados Reais (Observações)"
)  # Gráfico de dispersão[cite: 1]
plt.plot(
    X, y_pred_all, color="red", linewidth=2, label="Reta de Regressão"
)  # Linha de tendência[cite: 1]
plt.xlabel("Variável Explicativa (X)")
plt.ylabel("Variável Alvo (y)")
plt.title("Modelo de Regressão Linear Simples")
plt.legend()
plt.grid(True)
plt.show()  # Exibe o gráfico interativo[cite: 1]
```
